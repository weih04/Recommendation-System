## 2. How predictable is the next move?

Unconditional entropy of the next move at each ply, and the best a 1st-order Markov model could do on
its own training data (always predict the most common successor of the previous move). This is the
baseline GRU4Rec has to beat, and the entropy curve shows *where* a longer memory could help.

```python
pairs, by_pos = Counter(), {}
for toks in train["tokens"]:
    for i in range(1, len(toks)):
        if toks[i] == UNK: continue
        pairs[(toks[i-1], toks[i])] += 1
        by_pos.setdefault(min(i, 60), Counter())[toks[i]] += 1

def entropy(counter):
    p = np.array(list(counter.values()), dtype=float); p /= p.sum()
    return float(-(p * np.log2(p)).sum())

ent = pd.Series({pos: entropy(c) for pos, c in sorted(by_pos.items())})
ent.plot(figsize=(8, 3), title="Entropy of next move by ply (train)", color="#4C72B0")
plt.axvline(games["opening_ply"].median(), ls="--", c="grey", label="median opening_ply")
plt.xlabel("ply (capped at 60)"); plt.ylabel("bits"); plt.legend(); plt.tight_layout()
plt.savefig(IMAGES / "eda_3_entropy_by_ply.png", dpi=150, bbox_inches="tight"); plt.show()

succ = {}
for (a, b), c in pairs.items():
    succ.setdefault(a, Counter())[b] += c
hits = sum(max(c.values()) for c in succ.values())
print(f"1st-order Markov train-set HR@1 ceiling: {hits / sum(pairs.values()):.3%}")
print(f"in-book share of all plies: {long['in_book'].mean():.1%}  (median opening_ply = {games['opening_ply'].median():.0f})")
```

##### Why first order markov is insufficient...

Move predictability drops sharply once players leave standard opening theory. Uncertainty rises from 2.7 bits at ply 1 to a plateau of ~9 bits past ply 40, climbing most steeply around the median book exit at ply 4. With 92.1% of all moves falling outside book, relying only on the single previous move is insufficient: a first-order Markov chain caps out at just 11.2% HR@1, even under best-case training-set evaluation.


### Would a higher-order Markov chain be enough?

A first-order chain sees only the previous move. Extending the context to 2, 3 or 4 plies should help, but only where that exact sequence was seen during training. Each order is fitted on train and evaluated on **val**, reporting coverage (the share of val positions whose context exists in the
training data) separately from accuracy, plus a backoff variant that falls back to shorter contexts when the long one is missing.

```python
from collections import defaultdict

val = games[games["split"] == "val"]

def fit_markov(order, seqs):
    d = defaultdict(Counter)
    for toks in seqs:
        for i in range(order, len(toks)):
            if toks[i] == UNK: continue
            d[tuple(toks[i-order:i])][toks[i]] += 1
    return {ctx: c.most_common(1)[0][0] for ctx, c in d.items()}

ORDERS = [1, 2, 3, 4]
markov = {o: fit_markov(o, train["tokens"]) for o in ORDERS}
for o in ORDERS:
    print(f"order {o}: distinct contexts seen in train = {len(markov[o]):,}")

rows = []
for o in ORDERS:
    n = covered = hit = hit_backoff = 0
    for toks in val["tokens"]:
        for i in range(1, len(toks)):
            if toks[i] == UNK: continue
            n += 1
            pred = markov[o].get(tuple(toks[i-o:i])) if i >= o else None
            if pred is not None:
                covered += 1; hit += pred == toks[i]; hit_backoff += pred == toks[i]
            else:
                for k in range(min(o, i) - 1, 0, -1):          # back off to shorter contexts
                    p = markov[k].get(tuple(toks[i-k:i]))
                    if p is not None:
                        hit_backoff += p == toks[i]; break
    rows.append({"order": o, "val coverage": covered / n, "HR@1 | covered": hit / covered,
                 "HR@1 overall": hit / n, "HR@1 with backoff": hit_backoff / n})

order_results = pd.DataFrame(rows).set_index("order")
print(); print((order_results * 100).round(1).to_string())
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

order_results["val coverage"].plot.bar(ax=ax[0], color="#C44E52", rot=0,
                                       title="Share of val positions whose context was seen in train")
ax[0].set_ylabel("coverage"); ax[0].set_xlabel("Markov order")

order_results[["HR@1 | covered", "HR@1 overall", "HR@1 with backoff"]].plot(
    ax=ax[1], marker="o", title="Accuracy vs. context length (val)")
ax[1].set_ylabel("HR@1"); ax[1].set_xlabel("Markov order"); ax[1].set_xticks(ORDERS); ax[1].grid(alpha=.3)

plt.tight_layout(); plt.savefig(IMAGES / "eda_4_markov_order.png", dpi=150, bbox_inches="tight"); plt.show()

best = order_results["HR@1 with backoff"].idxmax()
print(f"best backoff order: {best}  ({order_results.loc[best, 'HR@1 with backoff']:.1%} HR@1 on val)")
```

Longer context is genuinely more informative but under the condition where the context exists in the training data,
accuracy rises steeply with order. Coverage collapses just as fast: chess sequences diverge so
quickly that a 4-ply context matches only a small minority of val positions, and the rest get no
prediction at all. Order 2 with backoff to order 1 is the best a counting model can do here.

This is the empirical case for GRU4Rec. The data says long context carries real signal that a
counting model cannot reach, because counting needs *exact* matches. A recurrent model generalises
across *similar* contexts through learned embeddings, so it can use long history without the
coverage cliff. It also fixes the baseline for the comparison: the Markov model to beat is
**order-2 with backoff**, not plain first-order.

### Is prediction a different problem in each phase?

Three phases per game, using the dataset's own book boundary rather than a fixed cut: *opening*
(`position < opening_ply`), *endgame* (the last 10 plies), *middlegame* (everything between).
For each, how many distinct items are in play and how well a 1st-order Markov model does there.
This is what justifies reporting the model comparison per phase later.

```python
def phases(toks, book):
    n = len(toks)
    end_start = max(book, n - 10)
    for i, t in enumerate(toks):
        yield ("opening" if i < book else "endgame" if i >= end_start else "middlegame"), i, t

phase_tokens = {p: Counter() for p in ["opening", "middlegame", "endgame"]}
phase_pairs = {p: {} for p in ["opening", "middlegame", "endgame"]}
for toks, book in zip(train["tokens"], train["opening_ply"]):
    tagged = list(phases(toks, book))
    for ph, i, t in tagged:
        phase_tokens[ph][t] += 1
        if i > 0 and t != UNK:
            phase_pairs[ph].setdefault(toks[i-1], Counter())[t] += 1

rows = []
for ph, cnt in phase_tokens.items():
    d = phase_pairs[ph]
    n = sum(sum(c.values()) for c in d.values())
    rows.append({"phase": ph, "plies": sum(cnt.values()), "distinct items": len(cnt),
                 "items covering 50%": (np.cumsum(sorted(cnt.values(), reverse=True)) < sum(cnt.values()) / 2).sum() + 1,
                 "markov HR@1": sum(max(c.values()) for c in d.values()) / n})
summary = pd.DataFrame(rows).set_index("phase")
print(summary.round(3).to_string())

print("\ntop 10 items per phase:")
print(pd.DataFrame({ph: [f"{t} ({c:,})" for t, c in cnt.most_common(10)] for ph, cnt in phase_tokens.items()}).to_string())
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
colors = {"opening": "#4C72B0", "middlegame": "#55A868", "endgame": "#8172B2"}

summary["distinct items"].plot.bar(ax=ax[0], color=[colors[p] for p in summary.index],
                                   title="Distinct items used per phase", rot=0)
ax[0].set_ylabel("items")

summary["markov HR@1"].plot.bar(ax=ax[1], color=[colors[p] for p in summary.index],
                                title="1st-order Markov HR@1 per phase", rot=0)
ax[1].set_ylabel("HR@1"); ax[1].axhline(summary["markov HR@1"].mean(), ls="--", c="grey", label="overall mean")
ax[1].legend()

for ph, cnt in phase_tokens.items():
    c = np.array(sorted(cnt.values(), reverse=True), dtype=float)
    ax[2].plot(np.arange(1, len(c) + 1), np.cumsum(c) / c.sum(), label=ph, color=colors[ph])
ax[2].set_xscale("log"); ax[2].set_xlabel("items (log, by rank)"); ax[2].set_ylabel("cumulative share of plies")
ax[2].set_title("Item concentration per phase"); ax[2].legend(); ax[2].grid(alpha=.3)

plt.tight_layout(); plt.savefig(IMAGES / "eda_6_phases.png", dpi=150, bbox_inches="tight"); plt.show()
```

Prediction difficulty is driven primarily by game phase. In the opening, players rely heavily on memorized theory: with only 233 distinct moves used, six of which cover half of all plies, a first-order Markov chain reaches 41.9% HR@1. However, this predictability vanishes in the middlegame (77% of all plies), where move variety explodes to 2,922 items and Markov accuracy collapses to 10.1%. The endgame is the hardest phase, with the lowest accuracy (9.4%) and the flattest move distribution (308 items to cover half its plies, versus 150 in the middlegame). Because endgame top moves are position-specific king maneuvers that rarely repeat across different games, a first-order Markov chain struggles to generalize.

This breakdown yields two key insights. First, evaluating models with a single overall HR@k hides a four-fold difference across phases and simply mirrors middlegame trends, hence evaluation metrics must be split by phase. Second, because the opening is already near the ceiling for simple counting models, any advantage from a sequential architecture like GRU4Rec must come from the middle and endgames, especially the endgame, where pattern generalization matters far more than memorizing repeat move pairs.

### The Markov model, drawn

Rows are the previous move, columns the next. Colour is `P(next | prev)` over the 25 most common items.
Bright isolated cells mean the previous move almost determines the next one; a flat row means the
previous move carries little information and a longer context should help.

```python
top_items = [t for t, _ in train_counts.most_common(26) if t != UNK][:25]
M = pd.DataFrame(0.0, index=top_items, columns=top_items)
for (a, b), c in pairs.items():
    if a in M.index and b in M.columns:
        M.loc[a, b] = c
M = M.div(M.sum(axis=1).replace(0, np.nan), axis=0).fillna(0)

plt.figure(figsize=(9, 7.5))
plt.imshow(M, cmap="Blues"); plt.colorbar(label="P(next | prev)")
plt.xticks(range(len(top_items)), top_items, rotation=90); plt.yticks(range(len(top_items)), top_items)
plt.xlabel("next move"); plt.ylabel("previous move")
plt.title("1st-order transition probabilities (25 most common items)")
plt.tight_layout(); plt.savefig(IMAGES / "eda_5_transition_heatmap.png", dpi=150, bbox_inches="tight"); plt.show()
```

## 3. Evaluation segments

Results will be broken down by rating band and opening family, but a breakdown is only worth
reporting if each group holds enough test games for the metric to be stable. HR@10 measured on a
few dozen games carries a confidence interval wide enough to swallow any difference between models.
This section checks those group sizes before they are committed to.

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3.5))
games["rating_band"].value_counts().reindex(BAND_ORDER).plot.bar(ax=ax[0], title="rating band", color="#4C72B0")
games["eco_letter"].value_counts().sort_index().plot.bar(ax=ax[1], title="ECO family", color="#55A868")
games["victory_status"].value_counts().plot.bar(ax=ax[2], title="victory status", color="#C44E52")
plt.tight_layout(); plt.savefig(IMAGES / "eda_7_segments.png", dpi=150, bbox_inches="tight"); plt.show()

test = games[games["split"] == "test"]
print("test games per rating band:"); print(test["rating_band"].value_counts().reindex(BAND_ORDER))
print(f"\ntest games with |rating_diff| < 200: {(test['rating_diff'].abs() < 200).sum()} of {len(test)}")
pd.crosstab(test["rating_band"], test["rating_diff"].abs() < 200).reindex(BAND_ORDER).rename(
    columns={True: "balanced", False: "mismatched"})
```

To ensure reliable metrics and clean data splits, four core decisions were finalized:
- Rating Band Merging: Due to small test set sizes, the extreme rating bands (<1200 with 33 games and 2200+ with 49 games) were merged into <1600 and 2000+, respectively. Raw, fine-grained bands remain preserved in the data artifact for future re-analysis.
- Rating Difference Filtering: Per-band metrics are restricted to games with $\vert{}rating\_diff\vert{} < 200$ (1,451 of 1,891 test games). Because rating bands are based on average player rating, filtering removes mismatched games where the label fits neither player, ensuring clean evaluation without sacrificing sample size. Unfiltered metrics are also provided for comparison.
- Opening Families: Opening categories A–D are reported directly, whereas Category E is flagged with its sample size (~490 games) due to volume.
- Timeout Wins: Games decided by running out of time (outoftime) are isolated as a separate slice when analyzing win labels, as these outcomes reflect clock pressure rather than move quality.

### Why the split is temporal

Games per month: the dataset is heavily skewed towards 2017, so the last 10% of games spans only a few
days. A random split would mix eras and let a player's later games predict their earlier ones.

## 4. Does `winner` carry a signal about move quality?

To evaluate whether game outcome (winner) serves as a reliable proxy for move quality, we examined move selection across board contexts, opening families, player ratings, and termination types.

```python
dec = games[games["winner"] != "draw"].copy()
dec["white_won"] = dec["winner"] == "white"

first = dec.assign(first_move=dec["tokens"].str[0]).groupby("first_move")["white_won"].agg(["mean", "size"])
print("white win rate by first move (n >= 100):")
print(first[first["size"] >= 100].sort_values("mean", ascending=False).round(3))
print("\nwhite win rate by ECO family:")
print(dec.groupby("eco_letter")["white_won"].agg(["mean", "size"]).round(3))
```

```python
# how much of 'winner' is just rating?
dec["stronger_won"] = np.sign(dec["rating_diff"]) == np.where(dec["white_won"], 1, -1)
gap = pd.cut(dec["rating_diff"].abs(), [-1, 50, 100, 200, 400, np.inf],
             labels=["0-50", "50-100", "100-200", "200-400", "400+"])
by_gap = dec.groupby(gap, observed=True)["stronger_won"].agg(["mean", "size"])
print(by_gap.round(3))

by_gap["mean"].plot.bar(figsize=(6, 3), color="#DD8452", edgecolor="white",
                        title="P(stronger player wins) by rating gap")
plt.axhline(0.5, ls="--", c="grey"); plt.ylabel("share"); plt.xlabel("|rating_diff|"); plt.tight_layout()
plt.savefig(IMAGES / "eda_9_rating_gap_vs_outcome.png", dpi=150, bbox_inches="tight"); plt.show()

print(f"\nvictory_status among decisive games:\n{dec['victory_status'].value_counts(normalize=True).round(3)}")
```

```python
# for the 20 most common 2-ply contexts: do winners and losers choose the next move differently?
ctx, ctx_w = Counter(), {}
for toks, w in zip(train["tokens"], train["winner"]):
    for i in range(2, len(toks)):
        key = (toks[i-2], toks[i-1]); ctx[key] += 1
        if w != "draw":
            side = "w" if i % 2 == 0 else "b"
            ctx_w.setdefault(key, {"win": Counter(), "loss": Counter()})["win" if side == w[0] else "loss"][toks[i]] += 1

rows = []
for key, _ in ctx.most_common(20):
    w, l = ctx_w[key]["win"], ctx_w[key]["loss"]
    rows.append({"context": " ".join(key), "winners_top": w.most_common(1)[0][0], "losers_top": l.most_common(1)[0][0],
                 "n_win": sum(w.values()), "n_loss": sum(l.values())})
div = pd.DataFrame(rows)
print(f"contexts where winners and losers differ on the top move: {(div['winners_top'] != div['losers_top']).sum()} of {len(div)}")
div
```

While game outcome (winner) is often used as a proxy for move quality, empirical analysis reveals it to be a weak and heavily confounded signal. Move choice and outcome are related (evidence: white's win rate ranges from 56.2% after c4 to 35.5% after e3). But in early play winners and losers make nearly identical choices, differing on the top move in only 2 of the 20 most common two-ply contexts, and win rates barely move across ECO families (50.4%–54.1%). 

The label is instead dominated by skill disparity, with the higher-rated player winning 85.9% of games at a 400+ rating gap versus 49.8% in balanced matchups, and by clock conditions, with 8.4% of decisive games ending in time forfeit. To ensure the winner label reflects superior play rather than rating gaps or clock pressure, winner-weighted training and evaluation are restricted to balanced matchups (|rating_diff| < 200) and exclude time-out victories. Because winners' and losers' opening choices are so similar, any resulting gain should be expected in the middlegame and endgame.

## 5. Player activity — why not a player-based model?

FPMC, matrix factorisation and other user-item models need users with a history. Combining `white_id`
and `black_id` gives the number of games per player in the raw data.

```python
player_games = (raw["white_id"].value_counts()
                .add(raw["black_id"].value_counts(), fill_value=0).astype(int).sort_values(ascending=False))

print(f"unique players: {len(player_games):,}")
print(player_games.describe().round(1))
for pct in [50, 75, 90, 95, 99]:
    print(f"  {pct}th percentile: {np.percentile(player_games.values, pct):.0f} games")

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].hist(player_games.values, bins=100, color="#DD8452", edgecolor="white", alpha=0.85)
axes[0].set_yscale("log"); axes[0].set_xlabel("games played"); axes[0].set_ylabel("players (log)")
axes[0].set_title("Player activity distribution")

sorted_vals = np.sort(player_games.values)
axes[1].plot(sorted_vals, np.arange(1, len(sorted_vals) + 1) / len(sorted_vals), color="#4C72B0", lw=2)
axes[1].axhline(0.5, c="grey", ls="--", alpha=.6); axes[1].axhline(0.9, c="grey", ls=":", alpha=.6)
axes[1].set_xlabel("games played"); axes[1].set_ylabel("cumulative share of players")
axes[1].set_title("CDF of player activity"); axes[1].set_xscale("log"); axes[1].grid(alpha=.3)

plt.tight_layout(); plt.savefig(IMAGES / "eda_10_player_activity.png", dpi=150, bbox_inches="tight"); plt.show()

for label, mask in [("exactly 1 game", player_games == 1), ("<= 3 games", player_games <= 3), ("> 10 games", player_games > 10)]:
    print(f"players with {label:>15}: {mask.sum():>6,} ({mask.mean():.1%})")
```

Player history is far too sparse to support a user-based model. Across 15,635 players the median player appears in exactly one game: 71.4% have a single game, 90.8% have three or fewer, and only 4.1% exceed ten (max 154). The CDF crosses 50% before the first data point and reaches 90% by three
games, after which it is essentially flat.
